```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/itsme/Onedrive/Desktop/CognifyzTechnologies_DataScience_Intership/Dataset .csv")
```

```python
print("Unique Price Ranges:")
print(df['Price range'].unique())
```

```python
price_labels = {
    1: 'Low',
    2: 'Medium',
    3: 'High',
    4: 'Luxury'
}
df['Price range label'] = df['Price range'].map(price_labels)
```

```python
price_range_counts = df['Price range label'].value_counts()
print("\nNumber of Restaurants in Each Price Range:")
print(price_range_counts)
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Price range label', data=df, palette='pastel', order=price_labels.values())
plt.title('Distribution of Restaurants by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')
plt.tight_layout()
plt.show()
```

```python
avg_rating_per_price = df.groupby('Price range label')['Aggregate rating'].mean().sort_values()
print("\nAverage Aggregate Rating by Price Range:")
print(avg_rating_per_price)
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x=avg_rating_per_price.index, y=avg_rating_per_price.values, palette='viridis')
plt.title('Average Aggregate Rating by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Average Rating')
plt.tight_layout()
plt.show()
```

```python
avg_votes_per_price = df.groupby('Price range label')['Votes'].mean().sort_values()
print("\nAverage Votes by Price Range:")
print(avg_votes_per_price)
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x=avg_votes_per_price.index, y=avg_votes_per_price.values, palette='coolwarm')
plt.title('Average Votes by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Average Votes')
plt.tight_layout()
plt.show()
```
