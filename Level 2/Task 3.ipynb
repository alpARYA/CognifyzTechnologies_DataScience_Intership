```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("C:/Users/itsme/Onedrive/Desktop/CognifyzTechnologies_DataScience_Intership/Dataset .csv")
```

```python
df.columns = df.columns.str.strip().str.lower()
```

```python
print("Columns in dataset:\n", df.columns.tolist())
```

```python
binary_columns = ['online order', 'book table']
for col in binary_columns:
    if col in df.columns:
        df[col] = df[col].map({'Yes': 1, 'No': 0})
```

```python
if 'cuisines' in df.columns:
    df['cuisine count'] = df['cuisines'].fillna('').apply(lambda x: len(x.split(',')))
```

```python
df['primary cuisine'] = df['cuisines'].fillna('').apply(lambda x: x.split(',')[0] if x else 'Unknown')
```

```python
if 'price range' in df.columns:
    df['price label'] = df['price range'].map({1: 'Low', 2: 'Medium', 3: 'High', 4: 'Luxury'})
```

```python
if 'location' in df.columns:
    le_location = LabelEncoder()
    df['location encoded'] = le_location.fit_transform(df['location'].astype(str))

le_cuisine = LabelEncoder()
df['primary cuisine encoded'] = le_cuisine.fit_transform(df['primary cuisine'].astype(str))
```

```python
if 'aggregate rating' in df.columns:
    df['aggregate rating'] = pd.to_numeric(df['aggregate rating'], errors='coerce')
```

```python
columns_to_show = [
    col for col in [
        'online order', 'book table', 'cuisine count', 'primary cuisine',
        'price label', 'location encoded', 'primary cuisine encoded', 'aggregate rating'
    ] if col in df.columns  # Show only available ones
]

print("\nFeature-Engineered Data Snapshot:")
print(df[columns_to_show].head())
```
