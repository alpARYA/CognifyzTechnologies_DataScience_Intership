```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/itsme/Onedrive/Desktop/CognifyzTechnologies_DataScience_Intership/Dataset .csv")
```

```python
print("Descriptive Statistics for Numerical Columns:\n")
print(df.describe())
```

```python
print("\nDataset Information:\n")
print(df.info())
```

```python
print("\nUnique value counts for categorical columns:\n")
categorical_cols = df.select_dtypes(include='object').columns
for col in categorical_cols:
    print(f"{col}: {df[col].nunique()} unique values")
    print(df[col].value_counts().head(), "\n")
```

```python
selected_columns = ['Aggregate rating', 'Rating color', 'Restaurant Name', 'Country Code']
for col in selected_columns:
    if col in df.columns:
        print(f"\nMost common values in '{col}':")
        print(df[col].value_counts().head())
```

```python
print("\nMissing Values Summary:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df['Aggregate rating'] = pd.to_numeric(df['Aggregate rating'], errors='coerce')
```

```python
print("Descriptive Statistics for Numerical Columns:\n")
print(df.describe())
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x='Aggregate rating', data=df, palette='coolwarm')
plt.title('Count of Aggregate Ratings')
plt.xlabel('Aggregate Rating')
plt.ylabel('Count')
plt.show()
```

```python
if 'Rating color' in df.columns:
    rating_color_counts = df['Rating color'].value_counts()
    plt.figure(figsize=(7,7))
    plt.pie(rating_color_counts, labels=rating_color_counts.index, autopct='%1.1f%%', colors=sns.color_palette('pastel'))
    plt.title('Rating Color Distribution')
    plt.show()
```

```python
if 'Country Code' in df.columns:
    country_counts = df['Country Code'].value_counts().head(10)
    plt.figure(figsize=(10,6))
    sns.barplot(x=country_counts.index, y=country_counts.values, palette='Set2')
    plt.title('Top 10 Country Codes by Record Count')
    plt.xlabel('Country Code')
    plt.ylabel('Number of Records')
    plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```
