```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/itsme/Onedrive/Desktop/CognifyzTechnologies_DataScience_Intership/Dataset .csv")
```

```python
print(df.head())
```

```python
geo_df = df.dropna(subset=['Latitude', 'Longitude'])
```

```python
geo_df['Latitude'] = pd.to_numeric(geo_df['Latitude'], errors='coerce')
geo_df['Longitude'] = pd.to_numeric(geo_df['Longitude'], errors='coerce')
```

```python
geo_df = geo_df.dropna(subset=['Latitude', 'Longitude'])
```

```python
map_center = [geo_df['Latitude'].median(), geo_df['Longitude'].median()]
restaurant_map = folium.Map(location=map_center, zoom_start=3)
```

```python
marker_cluster = MarkerCluster().add_to(restaurant_map)
```

```python
for idx, row in geo_df.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=row.get('Restaurant Name', 'Unnamed')
    ).add_to(marker_cluster)
```

```python
restaurant_map.save("restaurant_geospatial_map.html")
print("Map saved as 'restaurant_geospatial_map.html'. You can open it in a browser.")
```
