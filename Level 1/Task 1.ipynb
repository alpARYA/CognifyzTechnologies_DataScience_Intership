```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/itsme/Onedrive/Desktop/CognifyzTechnologies_DataScience_Intership/Dataset .csv")
```

```python
print("Shape of the dataset:")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
print("\nData types of columns:")
print(df.dtypes)
```

```python
if df['Aggregate rating'].dtype != 'float':
    df['Aggregate rating'] = pd.to_numeric(df['Aggregate rating'], errors='coerce')
```

```python
print("\nDistribution of 'Aggregate rating':")
print(df['Aggregate rating'].value_counts().sort_index())
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x='Aggregate rating', data=df, palette='viridis')
plt.title('Distribution of Aggregate Ratings')
plt.xlabel('Aggregate Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("\nClass distribution in percentages:")
print((df['Aggregate rating'].value_counts(normalize=True).sort_index() * 100).round(2))
```
